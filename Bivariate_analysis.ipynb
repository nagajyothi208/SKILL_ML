```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Load the dataset (assuming it's a CSV file)
df = pd.read_csv('placement_predict_50k Dataset (1).csv') # Corrected filename

#BIVARIATE ANALYSIS

#Scatter Plot
plt.figure(figsize=(6,4))
sns.scatterplot(
    data=df,
    x="CGPA",
    y="AttendancePercent"
)
plt.title("CGPA vs Attendance Percentage")
plt.show()
#Regression Plot
plt.figure(figsize=(6,4))
sns.regplot(
    data=df,
    x="CGPA",
    y="CodingTestScore"
)
plt.title("CGPA vs Coding Test Score")
plt.show()
#Box Plot
plt.figure(figsize=(6,4))
sns.boxplot(
    data=df,
    x="PlacementStatus",
    y="CGPA"
)
plt.title("CGPA vs Placement Status")
plt.show()
#Violin Plot
plt.figure(figsize=(6,4))
sns.violinplot(
    data=df,
    x="PlacementStatus",
    y="AttendancePercent"
)
plt.title("Attendance vs Placement Status")
plt.show()
#Gender vs Placement
plt.figure(figsize=(6,4))
sns.countplot(
    data=df,
    x="Gender",
    hue="PlacementStatus"
)
plt.title("Gender vs Placement Status")
plt.show()
#Stream vs Placement
plt.figure(figsize=(10,5))
sns.countplot(
    data=df,
    x="Stream",
    hue="PlacementStatus"
)
plt.xticks(rotation=45)
plt.title("Stream vs Placement Status")
plt.show()
```
