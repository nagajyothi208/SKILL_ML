```python
# Hospital Treatment Cost Analysis

import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# -------------------------------
# Dataset
# -------------------------------
data = {
    'Treatment_Days': [2, 4, 5, 6, 8],
    'Diagnostic_Tests': [3, 5, 6, 7, 9],
    'Treatment_Cost': [35, 52, 60, 74, 90]
}

df = pd.DataFrame(data)

# -------------------------------
# (a) Scatter Plot
# -------------------------------
fig = plt.figure(figsize=(8,6))
ax = fig.add_subplot(111, projection='3d')

ax.scatter(df['Treatment_Days'],
           df['Diagnostic_Tests'],
           df['Treatment_Cost'],
           color='blue',
           s=100)

ax.set_xlabel('Treatment Days (X1)')
ax.set_ylabel('Diagnostic Tests (X2)')
ax.set_zlabel('Treatment Cost (Y)')
ax.set_title('Treatment Days, Diagnostic Tests vs Treatment Cost')

plt.show()

# -------------------------------
# (b) Linear Regression Model
# -------------------------------
X = df[['Treatment_Days', 'Diagnostic_Tests']]
Y = df['Treatment_Cost']

model = LinearRegression()
model.fit(X, Y)

# Predictions
predicted_cost = model.predict(X)

print("Intercept (a):", model.intercept_)
print("Coefficients (b1, b2):", model.coef_)

print("\nActual vs Predicted Cost")
for actual, predicted in zip(Y, predicted_cost):
    print(f"Actual: {actual}  Predicted: {predicted:.2f}")

# Example Prediction
new_patient = [[7, 8]]  # 7 treatment days, 8 tests
predicted = model.predict(new_patient)
print("\nPredicted Treatment Cost for (7 Days, 8 Tests):",
      round(predicted[0], 2), "₹ Thousands")

# -------------------------------
# (c) Actual vs Predicted Graph
# -------------------------------
plt.figure(figsize=(8,5))
plt.plot(Y.values, marker='o', label='Actual Cost')
plt.plot(predicted_cost, marker='s', label='Predicted Cost')

plt.title('Actual vs Predicted Treatment Cost')
plt.xlabel('Patient')
plt.ylabel('Treatment Cost (₹ Thousands)')
plt.xticks(range(len(Y)), ['P1', 'P2', 'P3', 'P4', 'P5'])
plt.legend()
plt.grid(True)

plt.show()
```

```python
# Hospital ICU Patient Data Analysis

import pandas as pd
import numpy as np

# Create Dataset
data = {
    'Patient': ['P1', 'P2', 'P3', 'P4'],
    'Heart_Rate': [78, np.nan, 82, 250],
    'Blood_Pressure': [118, 115, np.nan, 120],
    'Oxygen_Saturation': [97, 96, 95, np.nan]
}

df = pd.DataFrame(data)

print("ICU Patient Dataset:\n")
print(df)

# --------------------------------------------------
# (a) Identify Missing Values
# --------------------------------------------------

print("\nMissing Values in Each Column:")
print(df.isnull().sum())

# --------------------------------------------------
# (b) Detect Abnormal Heart Rate Values
# Normal HR Range: 60 - 120 bpm
# --------------------------------------------------

abnormal_hr = df[(df['Heart_Rate'] < 60) | (df['Heart_Rate'] > 120)]

print("\nPatients with Abnormal Heart Rate:")
print(abnormal_hr[['Patient', 'Heart_Rate']])

# --------------------------------------------------
# (c) Summary Statistics for BP and O2
# --------------------------------------------------

print("\nBlood Pressure Statistics:")
print(df['Blood_Pressure'].describe())

print("\nOxygen Saturation Statistics:")
print(df['Oxygen_Saturation'].describe())

# --------------------------------------------------
# Fill Missing Values with Mean (Optional)
# --------------------------------------------------

df['Heart_Rate'].fillna(df['Heart_Rate'].mean(), inplace=True)
df['Blood_Pressure'].fillna(df['Blood_Pressure'].mean(), inplace=True)
df['Oxygen_Saturation'].fillna(df['Oxygen_Saturation'].mean(), inplace=True)

print("\nDataset After Handling Missing Values:")
print(df)
```
